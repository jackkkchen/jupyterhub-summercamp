# Linear Regression

In this notebook, we'll look at two examples of linear regression.

---

## Why Linear Regression?

📌 Equation:  
**y = mx + b**  
Where:
- `y` is the predicted weight
- `x` is the body length
- `m` is the slope (how weight increases with length)
- `b` is the intercept


```python
# Load libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

## Create a Dataset

We'll simulate animal data where body length (in cm) relates to weight (in kg).


```python
np.random.seed(0)

# Generate body lengths (cm) for animals (between 30 cm and 200 cm)
body_length = np.random.uniform(30, 200, 50)

# Simulate weight with noise: weight increases with length
noise = np.random.normal(0, 10, size=50)
weight = body_length * 0.8 + 5 + noise  # simple linear relation

# Create DataFrame
animals = pd.DataFrame({
    'body_length_cm': body_length,
    'weight_kg': weight
})

animals.head()
```

## Visualize the Relationship

Let’s see how body length and weight relate visually.


```python
plt.figure(figsize=(8,5))
plt.scatter(animals['body_length_cm'], animals['weight_kg'], color='green')
plt.title("Animal Body Length vs. Weight")
plt.xlabel("Body Length (cm)")
plt.ylabel("Weight (kg)")
plt.grid(True)
plt.show()
```

## Fit the Regression Model


```python
X = animals[['body_length_cm']]
y = animals['weight_kg']

model = LinearRegression()
model.fit(X, y)

slope = model.coef_[0]
intercept = model.intercept_

print(f"Model: weight = {slope:.2f} * length + {intercept:.2f}")
```

## Visualize the Model


```python
plt.figure(figsize=(8,5))
plt.scatter(animals['body_length_cm'], animals['weight_kg'], color='green', label='Data')
plt.plot(animals['body_length_cm'], model.predict(X), color='red', label='Regression Line')
plt.xlabel("Body Length (cm)")
plt.ylabel("Weight (kg)")
plt.title("Linear Regression: Predicting Animal Weight")
plt.legend()
plt.grid(True)
plt.show()
```

## Try Predicting!

How much might an animal weigh if its body is 100 cm long?


```python
length_input = 100  # Try other values!
predicted_weight = model.predict([[length_input]])

print(f"An animal with {length_input} cm body length is predicted to weigh {predicted_weight[0]:.2f} kg.")
```

We want to see which variable has the most impact on Yearly Amount Spent - this will be the variable we want to predict, our target variable.
Now we can split the data into training and test sets:

```python
x = df[['Avg. Session Length', 'Time on App',
       'Time on Website', 'Length of Membership']]
y = df[['Yearly Amount Spent']]
```

```python
from sklearn.model_selection import train_test_split
```

```python
# Use sklearn function to split. Set the random_state to get the same result each time
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=101)
```

```python
# Create an instance of a LinearRegression model
lm = LinearRegression()
```

```python
# Fit the model to our training data
lm.fit(x_train, y_train)
```

```python
# Now we can use our model to predict y values for our test set:
predicted = lm.predict(x_test)
```

```python
# Plot real vs. predicted test values:
plt.pyplot.scatter(y_test, predicted)
plt.pyplot.ylabel('Predicted')
plt.pyplot.xlabel('Y test')
```

Our predicted values should be well correlated with the actual values!

How can we tell which features are most important to our model? Let's look at the coefficients:

```python
pd.DataFrame(lm.coef_[0], x.columns, columns=['Coefficient'])
```
